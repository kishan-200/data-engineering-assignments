```python
spark.sql("""
DESCRIBE HISTORY lab_db.customers
""").show(truncate=False)
```

```python
spark.sql("""
SELECT
    version,
    timestamp,
    operation
FROM (
    DESCRIBE HISTORY lab_db.customers
)
ORDER BY version ASC
""").show(truncate=False)
```

```python
spark.sql("""
SELECT *
FROM lab_db.customers VERSION AS OF 0
""").show(truncate=False)
```

```python
spark.sql("""
CREATE OR REPLACE TABLE lab_db.customers_dev_shallow
SHALLOW CLONE lab_db.customers
""")
```

```python
spark.sql("""
SELECT *
FROM lab_db.customers_dev_shallow
ORDER BY customer_id
""").show(truncate=False)
```

```python
spark.sql("""
DESCRIBE DETAIL lab_db.customers_dev_shallow
""").show(truncate=False)
```

```python
spark.sql("""
CREATE OR REPLACE TABLE lab_db.customers_backup_deep
DEEP CLONE lab_db.customers
""")
```

```python
spark.sql("""
SELECT *
FROM lab_db.customers_backup_deep
ORDER BY customer_id
""").show(truncate=False)
```
